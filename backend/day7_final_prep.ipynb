```python
import importlib.util

def load_module(name, path):
    spec = importlib.util.spec_from_file_location(name, path)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return module

ml = load_module("ml_engine", "ml_engine.py")
offer = load_module("offer_engine", "offer_engine.py")

print("Both engines loaded!")
```

```python
print("=== Testing ml_engine edge cases ===\n")

# Test 1: Empty strings
result = ml.analyze_resume("", "")
print("Test 1 - Empty strings:")
print(result)

# Test 2: None values
result = ml.analyze_resume(None, None)
print("\nTest 2 - None values:")
print(result)

# Test 3: Very short input
result = ml.analyze_resume("python", "developer")
print("\nTest 3 - Very short input:")
print(result)

# Test 4: Numbers and symbols only
result = ml.analyze_resume("123 !!! ###", "$$$ 456 ***")
print("\nTest 4 - Numbers and symbols:")
print(result)

# Test 5: Real world test
resume = """
Senior Python Developer with 5 years experience in machine learning,
deep learning, tensorflow, keras, scikit-learn, pandas, numpy, SQL,
REST APIs, docker, kubernetes, git, agile development.
"""
jd = """
Looking for Python Developer with machine learning experience,
scikit-learn, pandas, tensorflow, SQL, REST APIs, docker and git.
"""
result = ml.analyze_resume(resume, jd)
print("\nTest 5 - Real world test:")
print(f"Score: {result['score']}%")
print(f"Level: {result['level']}")
print(f"Missing skills: {result['missing_skills']}")
```

```python
resume = """
Senior Python Developer with 5 years experience in machine learning,
deep learning, tensorflow, keras, scikit-learn, pandas, numpy, SQL,
REST APIs, docker, kubernetes, git, agile development.
"""
jd = """
Looking for Python Developer with machine learning experience,
scikit-learn, pandas, tensorflow, SQL, REST APIs, docker and git.
"""

cleaned_resume = ml.clean_text(resume)
cleaned_jd = ml.clean_text(jd)

print("Cleaned resume words:")
print(set(cleaned_resume.split()))
print("\nCleaned JD words:")
print(set(cleaned_jd.split()))
print("\nJD - Resume (missing):")
print(set(cleaned_jd.split()) - set(cleaned_resume.split()))
```

```python
print("=== Testing offer_engine edge cases ===\n")

# Test 1: Empty string
result = offer.analyze_offer_letter("")
print("Test 1 - Empty:")
print(f"Clauses: {result['total_clauses']}")

# Test 2: No risky clauses
result = offer.analyze_offer_letter("""
Employee will receive 12 days paid leave annually.
Salary will be credited on last working day of month.
Employee is entitled to medical insurance from day one.
""")
print("\nTest 2 - All normal clauses:")
print(f"High risk: {result['high_risk_count']}")
print(f"Medium risk: {result['medium_risk_count']}")

# Test 3: All risky clauses
result = offer.analyze_offer_letter("""
Employee must serve bond period of 2 years or pay Rs 1,00,000.
Employee shall not join any competitor for 1 year after leaving.
All work produced during employment belongs to the company.
Employee must serve 90 days notice period before resignation.
""")
print("\nTest 3 - All risky clauses:")
print(f"High risk: {result['high_risk_count']}")
print(f"Medium risk: {result['medium_risk_count']}")
```

```python
engines_code = '''
from ml_engine import analyze_resume, get_match_score, get_missing_skills
from offer_engine import analyze_offer_letter, analyze_clause

__all__ = [
    "analyze_resume",
    "get_match_score", 
    "get_missing_skills",
    "analyze_offer_letter",
    "analyze_clause"
]
'''

with open('engines.py', 'w') as f:
    f.write(engines_code)

print("engines.py saved!")
```
