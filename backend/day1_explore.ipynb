```python
import pandas as pd

df = pd.read_csv('Resume.csv')
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
# How many job categories are there?
print(df['Category'].value_counts())
```

```python
# Let's read one full resume to see what the text actually looks like
print(df['Resume_str'][0])
```

```python
# Check for any missing values
print(df.isnull().sum())
```

```python
# How many unique categories?
print("Total categories:", df['Category'].nunique())
print("\nCategory list:")
print(df['Category'].unique())
```

```python
# Drop what we don't need, keep only useful columns
df_clean = df[['Resume_str', 'Category']].copy()
df_clean.columns = ['resume_text', 'category']

# Save it
df_clean.to_csv('resumes_clean.csv', index=False)
print("Saved! Shape:", df_clean.shape)
```
